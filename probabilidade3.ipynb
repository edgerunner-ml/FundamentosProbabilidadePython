# Valor Esperado

Uma variável aleatória discreta tem resultados finitos. Por exemplo, o lançamento de um dado tem apenas seis possíveis resultados. Para variáveis aleatórias discretas, o valor esperado é a soma dos possíveis resultados ponderados por sua probabilidade. Para calcular o valor esperado, cada resultado é multiplicado por sua probabilidade, e os produtos são somados. Em termos simples, o valor esperado é um valor em torno do qual a probabilidade vai se concentrar quando você repetir o experimento.

*Esta fórmula representa o Valor Esperado (ou esperança matemática) de uma variável aleatória discreta $X$, que funciona como uma média ponderada de todos os resultados possíveis.   Para calcular o valor esperado $E(X)$, multiplica-se cada valor possível que a variável pode assumir ($x_i$) pela sua respetiva probabilidade de ocorrer ($p_i$) e somam-se todos esses produtos, desde o primeiro até ao último resultado ($k$). Na prática, indica o valor médio que se espera obter se a experiência for repetida um grande número de vezes.*

## Em Python:

```python
from scipy.stats import describe
describe([0,1]).mean
```

*Isso é o valor esperado de uma moeda justa, considerando cara = 0 e coroa = 1.*

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

n_flips = 2000
flips = np.random.choice([0, 1], size=n_flips)
cumulative_mean = np.cumsum(flips) / np.arange(1, n_flips + 1)

plt.figure(figsize=(8, 5))
plt.plot(np.arange(1, n_flips + 1), cumulative_mean)
plt.axhline(0.5, color="gray", linestyle="-", linewidth=1.5)
plt.text(
    n_flips + 6, 0.5, "0.5", va="center", ha="left", fontsize=12, color="#333333"
)

plt.title("Valor Esperado do Lançamento de uma Moeda Justa")
plt.xlabel("Lançamentos")
plt.ylabel("Média dos Lançamentos")
plt.xlim(-10, n_flips * 1.08)
plt.ylim(-0.02, 1.03)
plt.grid(True)
plt.subplots_adjust(right=0.92)

plt.show()
```

#### *O gráfico mostra que, no início, a média dos lançamentos oscila bastante ao redor de $0,5$, mas à medida que o número de tentativas cresce, ela se estabiliza cada vez mais perto do valor esperado teórico ($0,5$). É exatamente isso que a Lei dos Grandes Números afirma: quanto maior a amostra, mais a média observada se aproxima da média real da população.*

# Variância

A variância mede o quão concentrados ou espalhados em relação ao valor esperado estão os dados. A variância é o valor esperado do desvio ao quadrado de uma variável aleatória em relação ao seu valor esperado.

# Valor Esperado e Variância na Distribuição Binomial

$$\Large \begin{array}{l}
\text{Para } X \sim \text{Binomial}(n, p): \\
E(X) = n \times p \\
\text{Var}(X) = n \times p \times (1 - p)
\end{array}$$

*No caso particular da distribuição binomial, o valor esperado é o produto do número de lançamentos de moeda pela probabilidade de sair cara. A variância é o produto do valor esperado pela probabilidade de falha (não sair cara).*

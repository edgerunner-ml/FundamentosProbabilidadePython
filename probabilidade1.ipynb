# Fundamentos de Probabilidade com Python

### Probabilidade é a base de muitos dos métodos e modelos em ciência de dados. Métodos de inferência estatística usam dados para entender os processos que geram esses dados. Com probabilidade, conseguimos estudar as regularidades que surgem em fenômenos aleatórios.

## Lançamento de Moedas

Ao jogar uma moeda, há dois resultados possíveis: cara ou coroa. Esse experimento aleatório é chamado de **ensaio de Bernoulli**, em homenagem a Jacob Bernoulli.
Um ensaio de Bernoulli é aquele em que os resultados possíveis são binários: podem ser modelados como sucesso ou fracasso, sim ou não, ligado ou desligado.
Aqui, o experimento é o lançamento da moeda; os resultados possíveis são: - **Sucesso:** dar cara. - **Fracasso:** dar coroa.
Cada resultado é chamado de evento. Você atribui probabilidades aos eventos. Com uma moeda justa, você tem cinquenta por cento de chance de dar cara e cinquenta por cento de chance de dar coroa em cada lançamento.

## Lançando uma moeda em Python

Para simular os lançamentos, vamos usar o objeto `bernoulli` da biblioteca `scipy.stats` do Python.
Podemos chamar `bernoulli.rvs`, passando dois argumentos, p=0.5 é a probabilidade de sucesso (1) em uma tentativa (50%), ficando 1 - p (50%) para o fracasso (0), e size=1 define que a amostra sorteada terá 1 único valor (0 ou 1) em um array do NumPy.

```python
from scipy.stats import bernoulli
bernoulli.rvs(p=0.5, size=1)
```

No scipy.stats, rvs significa Random Variates. Ele gera amostras aleatórias a partir da distribuição de probabilidade escolhida — no caso, uma distribuição de Bernoulli.

## Lançando várias moedas em Python

Para dez lançamentos de moeda, usamos `size=10`. Podemos usar `sum()` para saber quantas caras saíram.
Em bernoulli.rvs(p=0.5, size=10), p=0.5 é a probabilidade de sucesso (1) em cada tentativa (50%), ficando 1 - p (50%) para o fracasso (0), e size=10 define o tamanho da amostra sorteada — aqui, um array com 10 valores.

```python
from scipy.stats import bernoulli
bernoulli.rvs(p=0.5, size=10)
sum(bernoulli.rvs(p=0.5, size=10))
```

Uma sequência de ensaios de Bernoulli independentes segue a **distribuição binomial**.
Então, em vez de usar o objeto `bernoulli` e somar os resultados, podemos usar a distribuição binomial e o objeto `binom`.

```python
from scipy.stats import binom
binom.rvs(n=10, p=0.5, size=1)
```

## Lançando várias moedas ao mesmo tempo em Python

Com binom.rvs(n=10, p=0.5, size=10), você está simulando 10 experimentos binomiais, onde cada experimento consiste em 10 tentativas (como lançar uma moeda 10 vezes), com probabilidade de sucesso p=0.5 (50%) em cada tentativa. O resultado será um array com 10 números (de 0 a 10), e cada número indica quantos sucessos aconteceram dentro de um bloco de 10 tentativas

```python
from scipy.stats import binom
binom.rvs(n=10, p=0.5, size=10)
```

## Simulando moedas viciadas

Para simular uma moeda viciada, basta alterar o valor de p para a probabilidade de sucesso (1) desejada (diferente de 0.5, entre 0 e 1). Por exemplo, com p=0.3, cada lançamento tem 30% de chance de sucesso (1) e 70% de fracasso (0), seja simulando lançamentos individuais com bernoulli.rvs ou a quantidade de sucessos em blocos de n lançamentos com binom.rvs.

```python
from scipy.stats import binom
binom.rvs(n=10, p=0.3, size=10)
```

## Semente do gerador aleatório

A semente de um gerador aleatório é o número inicial que define a sequência de números pseudoaleatórios produzida pelo computador. Usamos isso para garantir reprodutibilidade: com a mesma semente, simulações e sorteios se repetem exatamente iguais em outras execuções ou máquinas, facilitando testes e comparações.

### Há duas opções para configurar a semente do gerador aleatório

1. Usando o parâmetro random_state da função rvs()

```python
from scipy.stats import binom
binom.rvs(n=10, p=0.3, size=10, random_state=42)
```

2. Usando a função random.seed() do numpy

```python
import numpy as np
from scipy.stats import binom
np.random.seed(42)
binom.rvs(n=10, p=0.3, size=10)
```

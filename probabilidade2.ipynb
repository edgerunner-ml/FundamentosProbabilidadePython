# Funções de Massa e de Distribuição de Probabilidade

# Função Massa

Uma variável aleatória discreta tem um número finito de resultados possíveis. A função de massa de probabilidade permite calcular a probabilidade de obter um resultado específico para uma variável aleatória discreta. A função de massa de probabilidade binomial permite calcular a probabilidade de obter k caras em n lançamentos de moeda, com p de probabilidade de dar cara.

*A formula mostra a função massa de probabilidade (PMF) da distribuicao binomial, que calcula a probabilidade exata de ter k sucessos em n tentativas independentes de Bernoulli, cada uma com probabilidade p de sucesso. O termo n escolhe k indica de quantas maneiras diferentes esses k sucessos podem ocorrer ao longo das n tentativas, p elevado a k é a probabilidade de ocorrerem k sucessos e (1 - p) elevado a (n - k) é a probabilidade de ocorrerem os demais n - k fracassos.*

#### Na distribuição binomial:
- k é a quantidade exata de sucessos que você quer avaliar (um número inteiro de 0 até n),
- n é o total de tentativas independentes de Bernoulli em uma rodada,
- p é a probabilidade de sucesso em cada tentativa individual (um valor entre 0 e 1, deixando 1 - p para o fracasso).

Em resumo, n é quantas vezes você tenta, p é a chance de sucesso em cada tentativa e k é o número de sucessos de interesse.

## Em Python:

```python
from scipy.stats import binom
binom.pmf(k=2, n=10, p=0.5)
```

*Isso é a probabilidade de sair 2 caras após 10 lançamentos de uma moeda justa.*

# Função de Distribuição Acumulada

#### *Se você quiser calcular a probabilidade de obter k caras ou menos em n lançamentos, usa a função de distribuição binomial acumulada, que soma as probabilidades de obter k caras ou menos em n lançamentos usa-se a função de distribuição binomial acumulada. A função de distribuição binomial acumulada permite calcular a probabilidade cumulativa de obter k caras ou menos em n lançamentos de moeda, com p de probabilidade de dar cara.*

## Em Python:

```python
from scipy.stats import binom
binom.cdf(k=5, n=10, p=0.5)
```

*Isso é a probabilidade de sair 5 caras ou menos após 10 lançamentos de uma moeda justa.*

#### *E para calcular a probabilidade de sair mais de 5 caras após 10 lançamentos de uma moeda justa, usamos o método binom.sf(k, n, p) ("survival function"), que já faz $1 - \text{cdf}(k, n, p)$ de forma direta e com maior precisão numérica quando a cauda é muito pequena.*

```python
from scipy.stats import binom
print(binom.sf(k=5, n=10, p=0.5))
```

*Isso calcula $P(X > 5)$ em 10 lançamentos de uma moeda justa (p=0.5), ou seja, a probabilidade de sair estritamente mais de 5 caras (6, 7, 8, 9 ou 10 caras).*

#### Lembre-se de que:
- binom.pmf() calcula a probabilidade de obter exatamente k caras em n lançamentos de moeda.
- binom.cdf() calcula a probabilidade de obter k caras ou menos em n lançamentos de moeda.
- binom.sf() calcula a probabilidade de obter mais de k caras em n lançamentos de moeda.
